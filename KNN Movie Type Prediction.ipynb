```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
```

```python
#Create the dataset
# Training data
data = {
    "Rating": [8.0, 6.2, 7.2, 8.2],
    "Duration": [160, 170, 168, 155],
    "Type": ["Action", "Action", "Comedy", "Comedy"]
}

df = pd.DataFrame(data)

# Display dataset
print(df)
```

```python
#Separate features and target
# Features
X = df[["Rating", "Duration"]]

# Target
y = df["Type"]

print("Features:")
print(X)

print("\nTarget:")
print(y)
```

```python
# train the KNN model
# Create KNN model
knn = KNeighborsClassifier(n_neighbors=3)

# Train the model
knn.fit(X, y)

print("KNN model trained successfully!")
```

```python
#Predict the movie type
# New movie: Housefull
housefull = pd.DataFrame({
    "Rating": [7.4],
    "Duration": [114]
})

# Prediction
prediction = knn.predict(housefull)

print("Predicted type for Housefull:", prediction[0])
```
